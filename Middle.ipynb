```python
#find the last match played by virat kolhi in delhi 
ipl["all_player"] = ipl["Team1Players"] + ipl["Team2Players"]
ipl
```

```python
def did_virat_kolhi(Lists) :
    return 'V kolhi' in Lists
```

```python
ipl['did_player_virat']= ipl['all_player'].apply(did_virat_kolhi)
ipl[((ipl['City'] == 'Delhi') & (ipl['did_player_virat'] == True))].drop_duplicates(subset=['city' , 'did_virat_kolhi'],keep='last')
```
